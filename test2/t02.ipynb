```python
from PIL import Image
import numpy as np

# 1. 이미지 파일 불러오기
img = Image.open("data/five.png")

# 2. 흑백 이미지로 변환 (0~255)
img = img.convert("L")

# 3. 28×28 크기로 조정
img = img.resize((28, 28))

# 4. 이미지 데이터를 숫자 배열로 변환
data = np.array(img)

# 5. 색상 반전 (MNIST 형식으로 변환)
# 흰색 배경(255) → 검은색(0)
# 검은색 숫자(0) → 흰색(255)
data = 255 - data

# 6. 2차원 배열 출력 (28×28)
print("28×28 배열")
print(data)
```

```python
# 7. 1차원 배열로 변환 (784개)
flat_data = data.flatten()
flat_data2 = data.reshape(1,-1)  # 1차원 배열로 변환

print("\n1차원 배열")
print(flat_data.tolist())
```

```python
for i,v in enumerate(flat_data.tolist()):
    # 숫자를 3자리 띄워쓰기로 출력
    print('{:3}'.format(v),end=" ")  # 255컬러 이기에 3자리
    # 28번째 enter키 
    if i%28==0: print()   # 이미지크기 28 X 28
```

```python
from sklearn import model_selection, svm, metrics
import pandas as pd
# csv파일이 2개 있음.
train_csv = pd.read_csv("data/train.csv",header=None)  # 파일 읽어오기
tk_csv = pd.read_csv("data/t10k.csv",header=None)
def test(l):
 output=[]
 for i in l:
     output.append(float(i)/256)
 return output
# 모든 row, 1열 부터 모두 가져옴. 이미지의 데이터는 0-1사이의 숫자여야 함.
train_csv_data = list(map(test,train_csv.iloc[:,1:].values))
# 모든 row, 1열 부터 모두 가져옴.
tk_csv_data = list(map(test,tk_csv.iloc[:,1:].values))
# print(tk_csv_data)
train_csv_label = train_csv[0].values  # list타입으로 출력
tk_csv_label = tk_csv[0].values
clf = svm.SVC()
# fit에 들어갈 데이터는 0~1사이의 값이어야 함.
clf.fit(train_csv_data,train_csv_label)
predict = clf.predict(tk_csv_data[0:1])
```
